```python
import pandas as pd

# Load the CSV
ds = pd.read_csv("D:\\Datastream\\EU_DSCD_RAW_LISTS\\EU_DSCD.csv")

# Drop rows with null or empty DSCD values
ds = ds[ds["DSCD"].notnull() & (ds["DSCD"].astype(str).str.strip() != "")]

# Create list of DSCD values
ds_list = ds.DSCD.tolist()

# Batch the list
batch_size = 1000
ds_batched = pd.DataFrame([ds_list[i:i+batch_size] for i in range(0, len(ds_list), batch_size)]).transpose()
ds_batched.columns = [f"L#EU{i+1:03d}" for i in range(ds_batched.shape[1])]

# Export to Excel
ds_batched.to_excel("D:\\Datastream\\EU_DSCD_batched.xlsx", index=False)
```
