```python
import pandas as pd
import os

folder = f"D:/Datastream/Firmcharacteristics_Monthly/US/Paneldata"
starting_file = os.listdir(folder)[0]
df_firm_monthly = pd.read_csv(os.path.join(folder, starting_file))

for file in os.listdir(folder)[1:]:
    df_tmp = pd.read_csv(os.path.join(folder, file))
    df_firm_monthly = pd.merge(df_firm_monthly, df_tmp, on = ["Date", "DSCD"], how = "outer")

df_firm_monthly = df_firm_monthly.drop(columns = ["EPSISURDTE", "EPSISURSUE"])
```

```python
df_firm_monthly.to_parquet(os.path.join(folder, "monthly_firm_panel_data.parquet"))
```
