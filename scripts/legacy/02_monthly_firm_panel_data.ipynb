```python
import pandas as pd
import os

folder = f"D:/Datastream/Firmcharacteristics_Monthly/US/Paneldata"
# Only read the per-variable CSV files (sorted for a reproducible column order). Previously all files in the
# folder were read, including the parquet written below, so a second run failed.
csv_files = sorted(f for f in os.listdir(folder) if f.endswith(".csv"))
print(f"Merging {len(csv_files)} variable files: {csv_files}")

df_firm_monthly = pd.read_csv(os.path.join(folder, csv_files[0]), parse_dates=["Date"])

for file in csv_files[1:]:
    df_tmp = pd.read_csv(os.path.join(folder, file), parse_dates=["Date"])
    df_firm_monthly = pd.merge(df_firm_monthly, df_tmp, on = ["Date", "DSCD"], how = "outer")

df_firm_monthly = df_firm_monthly.drop(columns = ["EPSISURDTE", "EPSISURSUE"], errors="ignore")
```

```python
df_firm_monthly.to_parquet(os.path.join(folder, "monthly_firm_panel_data.parquet"))
```
